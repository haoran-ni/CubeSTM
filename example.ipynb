```python
from cubestm import CubeSTM
from os.path import join
```

```python
dir = './data'
cube_path = join(dir, 'naph_cu_z6.cube')
geo_path = join(dir, 'naph_cu_z6.in')
instance = CubeSTM(cube_path, geo_path)
```

```python
instance.plot(show_atoms=['C', 'H'],
              canvas_height=60,
              canvas_width=60,
              neighbourhood=3,
              xy_decay=0.8,
              z_decay=1.,
              screening_power=10,
              localization_power=5,
              vmin=1e-2,
              vmax=1.0,
              sigma=2
              )
```
